```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("Apple is looking to buying UK startup for $1 billion dollars.")
```

```python
print("Whole doc:", doc.text)
```

```python
print("Tokens: ", [token.text for token in doc])
```

```python
span = doc[1:4] # slice of the document
print("Span: ", span.text)
```

```python
# Tokenization, Lemma, Stopwords

doc = nlp("Running runners run quickly, but better results come from training. Runner ran 1 mile in 4 minutes and twenty seconds.")

for token in doc:
  print(f"{token}, lemma={token.lemma_}, stopword={token.is_stop}, isdigit={token.like_num}")
```

```python
# Named Entity Recognition (NER)

doc = nlp("Google is opening a new office in Berlin on July 15, 2025. Azam is hiking on Rocky Mountain Colorado. Estes Park, CO just did a renovation for 20 million.")

for ent in doc.ents:
  print(ent.text, ent.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```
